```python
import os
import openai

client = openai.OpenAI(
    base_url="https://api.groq.com/openai/v1",
    api_key=os.environ.get("GROQ_API_KEY")
)
```

```python
import pandas as pd
import json
import re

def carregar_questoes_csv(caminho_arquivo):
    df = pd.read_csv(caminho_arquivo)
    questoes = []
    
    for _, row in df.iterrows():
        questao = {
            "id": row['numero'],
            "question": row['pergunta'],
            "alternatives": row['opcoes']
        }
        questoes.append(questao)
       
    return questoes

def obter_respostas(questoes):
    respostas = []
    
    for questao in questoes:
        prompt = f"""
        Questão {questao['id']}: {questao['question']}
        
        Opções:
        {questao['alternatives']}
        
        Qual é a resposta correta? Responda apenas com o texto da alternativa correspondente.
        """
        
        resposta = client.chat.completions.create(
            model="llama-3.3-70b-versatile",  # Ajuste para o modelo correto disponível na Groq
            messages=[{"role": "user", "content": prompt}]
        )

        resposta_texto = resposta.choices[0].message.content.strip()
        respostas.append({"id": questao["id"], "pergunta": questao["question"], "resposta_correta": resposta_texto})
        #criar um arquivo jsonl com a coluna 'label' com a respectiva resposta da api
    return respostas
```

```python
# Carregar questões do CSV
caminho_arquivo = "/kaggle/input/poscomp-24-bs/poscomp_24_b1.csv"  # Substitua pelo caminho do seu arquivo JSON
questoes = carregar_questoes_csv(caminho_arquivo)

# Obter respostas
llama_respostas_b1 = obter_respostas(questoes)

# Exibir resultados
for item in llama_respostas_b1:
    print(f"Pergunta: {item['pergunta']}")
    print('-------------------')
    print(f"Resposta Correta: {item['resposta_correta']}\n")
    print('-------------------')
    print(' ')
```
